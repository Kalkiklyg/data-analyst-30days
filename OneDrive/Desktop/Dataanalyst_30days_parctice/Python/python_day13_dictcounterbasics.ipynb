```python
# day13_dict_counter.py
from collections import Counter
feedbacks = ["good product", "bad packaging", "good support", "excellent", "bad quality"]
# word freq
words = " ".join(feedbacks).lower().split()
cnt = Counter(words)
print("top 5:", cnt.most_common(5))
# dict example
d = {"cust_1": 2, "cust_2": 5}
d["cust_3"] = 1
print("dict:", d)
```

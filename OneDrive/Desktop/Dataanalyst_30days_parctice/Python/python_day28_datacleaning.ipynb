```python
## day28_numpy_pandas_data_cleaning.py
import pandas as pd
import numpy as np

# 1. Load the CSV
df = pd.read_csv(r"C:/Users/Lenovo/Downloads/archive (19)/ASR.csv")

print("Raw data:\n", df.head())

```

```python
# Basic info
print("\nInfo:")
print(df.info())

#  Handle missing values
# Fill missing 'discount' with 0 and 'price' with column mean
df['Qty'] = df['Qty'].fillna(0)
df['Amount'] = df['Amount'].fillna(df['Amount'].mean())

#  Remove duplicates (if any)
df = df.drop_duplicates()


#  Use NumPy: classify high-value vs low-value orders
df['Order Id'] = np.where(df['Amount'] > 500, "High Value", "Low Value")

print("\nCleaned Data:\n", df)

#  Save cleaned version
df.to_csv("data/sales_cleaned.csv", index=False)
```

```python
# day27_integration_summary.py
import csv
from collections import Counter, defaultdict
import pandas as pd
import re

def parse_feedback(path="data/feedback.txt"):
    fb_by_cust = defaultdict(list)
    with open(path, encoding="utf-8") as f:
        for line in f:
            if ":" in line:
                cust, text = line.split(":", 1)
                fb_by_cust[cust.strip()].append(text.strip())
    return fb_by_cust

def clean_words(text):
    text = re.sub(r'[^\w\s]', '', text.lower())
    return text.split()

if __name__ == "__main__":
    fb = parse_feedback("data/feedback.txt")
    tx = pd.read_csv("data/transactions.csv")
    totals = tx.groupby("cust_id")["amount"].sum()
    summary = []
    for cust, texts in fb.items():
        all_words = []
        for t in texts:
            all_words += clean_words(t)
        top2 = Counter(all_words).most_common(2)
        total_spent = totals.get(cust, 0)
        summary.append((cust, total_spent, len(texts), top2))
    # print summary sorted by total_spent desc
    print(sorted(summary, key=lambda x: x[1], reverse=True))
```

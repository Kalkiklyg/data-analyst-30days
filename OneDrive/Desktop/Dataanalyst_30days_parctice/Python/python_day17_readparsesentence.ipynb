```python

# day17_file_io_parse_feedback.py
def parse_feedback(path):
    rows = []
    with open(path, encoding="utf-8") as f:
        for line in f:
            if ":" in line:
                cust, text = line.split(":", 1)
                rows.append((cust.strip(), text.strip()))
    return rows

if __name__ == "__main__":
    # assumes data/feedback.txt exists
    print(parse_feedback("data/feedback.txt"))
```

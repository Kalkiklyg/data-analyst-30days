```python
# day13_functions.py
def is_palindrome(s):
    s2 = "".join(ch for ch in s.lower() if ch.isalnum())
    return s2 == s2[::-1]

def count_vowels(s):
    return sum(1 for ch in s if ch.lower() in "aeiou")

print(is_palindrome("A man, a plan, a canal: Panama"))
print(count_vowels("hello world"))
```

```python
# day15_classes_customer.py
class Customer:
    def __init__(self, cust_id):
        self.cust_id = cust_id
        self.transactions = []
        self.feedbacks = []

    def add_transaction(self, amt):
        self.transactions.append(float(amt))

    def total_spent(self):
        return sum(self.transactions)

    def avg_spent(self):
        return self.total_spent() / len(self.transactions) if self.transactions else 0

cust = Customer("cust_1")
cust.add_transaction(499)
cust.add_transaction(1299)
print(cust.cust_id, cust.total_spent(), cust.avg_spent())
```

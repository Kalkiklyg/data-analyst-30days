```python
# day11_lists_slicing.py
lst = [5, 3, 9, 3, 1]
lst.append(7)
lst.extend([8, 9])
print("list:", lst)
print("slice [1:5]:", lst[1:5])
# deduplicate but keep order
seen = set()
dedup = []
for x in lst:
    if x not in seen:
        dedup.append(x)
        seen.add(x)
print("dedup:", dedup)
```
